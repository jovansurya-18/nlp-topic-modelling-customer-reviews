```python
# Number of common locations between the Google and Trustpilot data sets

g_locations = set(g_df["Club's Name"].dropna().str.lower().str.strip().unique())
tp_locations = set(tp_df["Location Name"].dropna().str.lower().str.strip().unique())

common_locations = g_locations.intersection(tp_locations)

print(f"Number of common locations between Google and Trustpilot datasets: {len(common_locations)}")
```

```python
# Data preprocessing – change to lower case,  remove numbers, remove stopwords, and tokenise using NLTK

import nltk
import re
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download("stopwords")
nltk.download("punkt_tab")

stop_words = set(stopwords.words("english"))

# Add domain-specific stopwords
custom_stopwords = {"gym", "pure", "puregym"}
stop_words.update(custom_stopwords)

def preprocess_text(text):
  # Lowercase
  text = text.lower()

  # Remove numbers
  text = re.sub(r'\d+', '', text)

  # Remove punctuation
  text = re.sub(r'[^\w\s]', '', text)

  # Tokenize
  tokens = word_tokenize(text)

  # Remove stopwords
  tokens = [word for word in tokens if word not in stop_words]

  return " ".join(tokens)

g_df["Preprocessed_comment"] = g_df["Comment"].apply(preprocess_text)
tp_df["Preprocessed_review"] = tp_df["Review Content"].apply(preprocess_text)
```

```python
# Find the frequency distribution of the words from each data set's reviews separately

from nltk.probability import FreqDist
import matplotlib.pyplot as plt

# Tokenise it again
g_df["tokens"] = g_df["Preprocessed_comment"].apply(word_tokenize)
tp_df["tokens"] = tp_df["Preprocessed_review"].apply(word_tokenize)

# Flatten
g_all_tokens = [word for tokens in g_df["tokens"] for word in tokens]
tp_all_tokens = [word for tokens in tp_df["tokens"] for word in tokens]

g_freq_dist = FreqDist(g_all_tokens)
tp_freq_dist = FreqDist(tp_all_tokens)

# Find top 10 words from each data set
g_top10 = g_freq_dist.most_common(10)
tp_top10 = tp_freq_dist.most_common(10)

# Split words and counts
g_words, g_counts = zip(*g_top10)
tp_words, tp_counts = zip(*tp_top10)

# Visualise
plt.figure(figsize=(10,10))
plt.bar(g_words, g_counts)
plt.title("Top 10 Words - Google Comments")
plt.xticks(rotation=45)
plt.xlabel("Words")
plt.ylabel("Counts")
plt.tight_layout()
plt.show()

plt.figure(figsize=(10,10))
plt.bar(tp_words, tp_counts)
plt.title("Top 10 Words - Trustpilot Reviews")
plt.xticks(rotation=45)
plt.xlabel("Words")
plt.ylabel("Counts")
plt.tight_layout()
plt.show()
```

# Conducting initial topic modelling

```python
# Filter out the reviews that are from the locations common to both data sets
g_neg_df["clean_location"] = g_neg_df["Club's Name"].str.lower().str.strip()
tp_neg_df["clean_location"] = tp_neg_df["Location Name"].str.lower().str.strip()

g_neg_locations = set(g_neg_df["clean_location"].dropna())
tp_neg_locations = set(tp_neg_df["clean_location"].dropna())

common_neg_locations = g_neg_locations.intersection(tp_neg_locations)

g_neg_common = g_neg_df[g_neg_df["clean_location"].isin(common_neg_locations)].copy()
tp_neg_common = tp_neg_df[tp_neg_df["clean_location"].isin(common_neg_locations)].copy()

# Use the raw reviews
g_neg_common = g_neg_common.rename(columns={"Comment": "review"})
tp_neg_common = tp_neg_common.rename(columns={"Review Content": "review"})


# Merge the reviews to form a new list
combined_neg_reviews = pd.concat([
    g_neg_common[["clean_location", "review"]],
    tp_neg_common[["clean_location", "review"]]
], ignore_index=True)

print(combined_neg_reviews.shape)
combined_neg_reviews.head()
```

```python
# Prepare for BERTopic

# Pass it into docs
docs = combined_neg_reviews["review"].dropna().tolist()

# Remove empty strings, just in case
docs = [doc for doc in docs if doc.strip() != ""]
```

```python
# Install BERTopic

!pip install bertopic
```

```python
from bertopic import BERTopic
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

# Add domain specific stopwords
custom_stopwords = list(ENGLISH_STOP_WORDS) + ["gym", "pure", "puregym"]

# Use CountVectorizer to allow full rich embedding while still filtering our keyword extraction
vectorizer = CountVectorizer(stop_words=custom_stopwords)

topic_model = BERTopic(vectorizer_model=vectorizer, nr_topics = 15)

topics, probs = topic_model.fit_transform(docs)

# Top topics with their document frequencies
topic_info = topic_model.get_topic_info()

# Remove outliers (-1)
topic_info = topic_info[topic_info["Topic"] != -1]

topic_info = topic_info.sort_values(
    by="Count",
    ascending=False
)

topic_info
```

```python
# List out top words from top 2 topics

top_2_topics = topic_info.head(2)["Topic"].tolist()

for topic in top_2_topics:
    print(f"Topic {topic}:")
    print(topic_model.get_topic(topic))
```

```python
# Load Falcon LLM

from transformers import pipeline

text_gen_pipeline = pipeline(
    "text-generation",
    model="tiiuae/falcon-7b-instruct",
    max_length=1000,
    device_map="auto"
)
```

```python
# Run prompt on all negative reviews from both data sets

# Combine data sets
combined_neg_reviews = pd.concat([
    g_neg_df["Comment"],
    tp_neg_df["Review Content"]
], ignore_index=True).dropna()

# Use a sample as runtime will expire
sample_reviews = combined_neg_reviews.sample(100, random_state=42)

# Prompt
prompt_template ="""
In the following customer review, pick out the main 3 topics.
Return them in a numbered list format, with each one on a new line.

Review:
{}
"""
```

```python
# Run the model

results = []

for review in sample_reviews:
    prompt = prompt_template.format(review)
    output = text_gen_pipeline(prompt)
    results.append(output[0]["generated_text"])
```

```python
# Extract the topics from the results

topics_list = []

for res in results:
    lines = res.split("\n")

    for line in lines:
        line = line.strip()

        # Keep only numbered lines (1., 2., 3.)
        if line.startswith(("1", "2", "3")):
            topic = line.split(".", 1)[-1].strip()
            topics_list.append(topic)

topics_list
```

```python
# Run BERTopic again

topic_model_llm = BERTopic(nr_topics=15)

topic_model_llm.fit_transform(topics_list)

topic_info_llm = topic_model_llm.get_topic_info()

# Remove outliers
topic_info_llm = topic_info_llm[topic_info_llm["Topic"] != -1]

# Sort
topic_info_llm = topic_info_llm.sort_values(
    by="Count",
    ascending=False
)

topic_info_llm
```

```python
# Bar chart

topic_model_llm.visualize_barchart(top_n_topics=10, n_words=5)
```

**Top 10 Topic Word Scores Bar Chart (First & Second & Third vs Fourth BERTopic run)**

The fourth run of BERTopic, which incorporates Falcon LLM-generated topic guidance, shows a clear improvement over the previous runs in terms of interpretability and insight quality. While the first three runs primarily produced keyword-based clusters that required manual interpretation, the fourth run generates more structured and meaningful semantic themes such as customer service issues, cleanliness and hygiene, pricing concerns, and service delivery problems.

Another improvement is the clearer separation and consolidation of topics. Previously, similar issues such as equipment faults or cleanliness were fragmented across multiple clusters, whereas in the fourth run they are grouped into more coherent and generalisable categories. Additionally, noise in the form of less meaningful words has been reduced, resulting in cleaner and more focused topics.

Additionally, new themes such as pricing and value perception emerge more explicitly, although they appear less dominant compared to operational issues. Overall, the results are more interpretable and actionable, as the topics align more closely with real business problem areas and reduce the need for manual interpretation.
